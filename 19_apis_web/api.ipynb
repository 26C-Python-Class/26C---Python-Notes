## Section 1: The Anatomy of a Request
In this section, we use the `requests` library to "knock on the door" of a server.

**Key Concepts:** Endpoints, Status Codes, and Headers.

```python
# Task 1: Your first API Call
import requests

# We use JSONPlaceholder - a free fake API for testing
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url)

print(f"Status Code: {response.status_code}") # Should be 200
print(f"Headers: {response.headers['Content-Type']}")
```



---

## Section 2: Data Translation (JSON Serialization)
APIs speak **JSON**, but Python speaks **Dictionaries**. Learning to translate between them is the core of API work.

**Key Concepts:** `json.loads()` (Input) and `json.dumps()` (Output).

```python
import json

# Task 2: Handling the "Payload"
raw_data = response.text  # This is a STRING
print(f"Type of raw_data: {type(raw_data)}")

# DESERIALIZATION: Convert JSON String to Python Dict
data_dict = json.loads(raw_data)
print(f"Post Title: {data_dict['title']}")

# SERIALIZATION: Convert Dict back to a "Pretty" JSON String for documentation
pretty_json = json.dumps(data_dict, indent=4)
print(pretty_json)
```

---

## Section 3: Authentication & Parameters
Most professional APIs (like Google Maps or Jumia) aren't wide open. They require an **API Key** or specific **Query Parameters**.

**Key Concepts:** Query Strings (`?key=value`) and Private Keys.

```python
# Task 3: Searching with Parameters
# We will use the 'filter' parameter to find posts by a specific User ID
search_url = "https://jsonplaceholder.typicode.com/posts"
query_params = {"userId": 1}

response = requests.get(search_url, params=query_params)
posts = response.json()

print(f"Found {len(posts)} posts for User 1.")
print(f"First Post Title: {posts[0]['title']}")
```

## Section 4: The Final Project — Live Weather App
Now, we combine everything into a functional tool using the **Open-Meteo API** (No key required for this demo).

**Challenge:** Write a function that takes a City's Latitude/Longitude and returns the current temperature.

```python
def get_weather(lat, lon):
    url = "https://api.open-meteo.com/v1/forecast"
    params = {
        "latitude": lat,
        "longitude": lon,
        "current_weather": True
    }
    
    response = requests.get(url, params=params)
    
    if response.status_code == 200:
        data = response.json()
        temp = data['current_weather']['temperature']
        return f"The current temperature is {temp}°C"
    else:
        return "Error: Could not reach the weather service."

# Test for Nairobi (Lat: -1.28, Lon: 36.82)
print(get_weather(-1.28, 36.82))
```

## Section 5: The "Unhappy Path" (Error Handling)
A Lead Architect always plans for failure. What happens if the internet goes down or the URL is wrong?

```python
# Task 5: Robust Error Handling
bad_url = "https://jsonplaceholder.typicode.com/this-page-does-not-exist"

try:
    res = requests.get(bad_url)
    res.raise_for_status() # This triggers an error if status is not 200
except requests.exceptions.HTTPError as err:
    print(f"HTTP Error Occurred: {err}")
except Exception as err:
    print(f"An unexpected error occurred: {err}")
```
